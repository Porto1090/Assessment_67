# Preprocessing in order to generate Dataset

```python
import os
import random

from pathlib import Path
ROOT = Path(os.getcwd()).parents[0]
```

```python
path_train_images = f"{ROOT}\\steganography\\dataset\\raw_train\\"
path_val_images = f"{ROOT}\\steganography\\dataset\\raw_val\\"
path_test_images = f"{ROOT}\\steganography\\dataset\\raw_test\\"
```

# Obteniendo lista de imagenes

```python
# Function to get a sorted list of images files in a given route

def get_images_list(images_route):
    # List of all images in the route
    images_list = os.listdir(images_route)
    images_list.sort()
    
    # print(len(images_list))
    
    # for i in range(len(images_list)):
      #   print(images_list[i])

    return images_list
```

```python
train_list = get_images_list(path_train_images)
print(len(train_list))
```

```python
val_list = get_images_list(path_val_images)
print(len(val_list))
```

```python
test_list = get_images_list(path_test_images)
print(len(test_list))
```

# Borrando imágenes sobrantes

```python
def delete_random_files(dict_route, images_list, wanted_amount):

    if wanted_amount >= len(images_list):
        print(f"Error: Solicitaste eliminar {wanted_amount} archivos, pero solo hay {len(images_list)} disponibles!")
    else:
        deleting_amount = len(images_list) - wanted_amount
        random_files = random.sample(images_list, deleting_amount)

        for file in random_files:

            complete_route = os.path.join(dict_route, file)
            os.remove(complete_route)

            print(f"Eliminado: {file}")
            
        print(f"\nSe eliminaron {deleting_amount} archivos aleatorios.")
```

```python
wanted_amount_train = 6400
delete_random_files(path_train_images, train_list, wanted_amount_train)
```

```python
train_list = get_images_list(path_train_images)
print(len(train_list))
```

```python
wanted_amount_val = 1600
delete_random_files(path_val_images, val_list, wanted_amount_val)
```

```python
val_list = get_images_list(path_val_images)
print(len(val_list))
```

```python
wanted_amount_test = 2000
delete_random_files(path_test_images, test_list, wanted_amount_test)
```

```python
test_list = get_images_list(path_test_images)
print(len(test_list))
```

# Renombrar archivos

```python
def rename_files(dict_route, files_list, set = None):

    for index, file_name in enumerate(files_list, start = 1):

        # Separar el nombre actual de su extensión ('foto.jpg' -> ('foto', '.jpg'))
        _, extension = os.path.splitext(file_name)
        
        # Crear el nuevo nombre
        new_name = f"{set}_{index}{extension}"
        
        # Rutas completas
        ruta_antigua = os.path.join(dict_route, file_name)
        ruta_nueva = os.path.join(dict_route, new_name)
        
        # Renombrar en el sistema
        os.rename(ruta_antigua, ruta_nueva)

    print(f"\nSe renombraron {len(files_list)} archivos del 1 al {len(files_list)}.")
```
